# <center>Input and Output quiz</center>

1. Write a program that reads a list of items, their prices, and quantities from a file. The input file should have each line in the format: item_name, price, quantity. (See example of input file)
2. The program should calculate the total amount for each item and the overall total.
3. Print an invoice with formatted string literals.
4. Save the invoice to a new file.

**Example of input file:**  
apple, 50.99, 50  
bananas, 23.50, 30  
oranges, 35.24, 100  

**Example of output:**
| Item    |   Price | Quantity |     Total |
|---------|---------|----------|-----------|
| apple   |   50.99 |       50 |   2549.50 |
| bananas |   23.50 |       30 |    705.00 |
| oranges |   35.24 |      100 |   3524.00 |
|---------|---------|----------|-----------|
| **Total** |         |          | **6778.50** |

**Percentage assignment:**
1. Creating a file with the specific format. [10 points]
2. Reading from the file. [20 points]
3. Calculating total amount for each item. [20 points]
4. Calculating the overall total. [10 points]
5. Formatting the output using formatted string literals. [25 points]
6. Saving the output to a new file. [15 points]

```python
with open('test.txt', 'r') as f:
  invoice = []
  for line in f:
    item = line.strip().split(", ")
    item_with_total = []
    product = item[0]
    price = float(item[1])
    quantity = int(item[2])
    total = price * quantity
    item_with_total.append(product)
    item_with_total.append(price)
    item_with_total.append(quantity)
    item_with_total.append(total)
    invoice.append(item_with_total)
  print(invoice)

with open('invoice.txt', 'w') as f:
  f.write('{:^10}{:>10}{:>10}{:>10}\n'.format('Item', 'Price', 'Quantity', 'Total'))
  overall = 0
  for e in invoice:
    overall += e[3]
    f.write('{:<10}{:>10.2f}{:>10}{:>10.2f}\n'.format(e[0], e[1], e[2], e[3]))
  f.write('----------------------------------------\n') 
  f.write('Total \t\t\t\t\t\t\t\t\t {:,}'.format(overall))

  


  # information = f.readlines()
  # invoice = []
  #   
  #   price = float(element[1])
  #   quantity = int(element[2])
  #   total = price * quantity
  #   element_with_total.append(element[0])
  #   element_with_total.append(price)
  #   element_with_total.append(quantity)
  #   element_with_total.append(total)
  #   invoice.append(element_with_total)
  # print(invoice)

```
