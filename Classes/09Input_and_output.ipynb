Often, we want more control over the formatting of our output than simply printing space-separated values. There are several ways to format output in Python:

1. Using formatted string literals.
2. Using the `format()` method.
3. Manual string formatting.

## 1. Formatted string literals

* To use **formatted string literals**, begin a string with `f` or `F` before the opening quotation mark or double quotation mark. 
* Inside this string, you can write a Python expression betwen `{` and `}` characters that can refer to variables or literal values.

**Example 1. Not using formatted string literals**

```python
part = 25
total = 200
print(part, "% of", total, "=", part * total / 100)
```

**Example 2. Computing percentage with formatted string literals**

```python
part = 25
total = 200
print(f'{part}% of {total} = {part * total / 100}')
```

**Example 3. Useful information about a student**

```python
name = "Valentina"
last_name = "Martinez"
age = 16
var = "friends"
print(f'Hello {var}, my name is {name} {last_name} and my age is {age}')
#print("Hello my name is", name, last_name, "and my age is", age)
```

**Example 4. Printing the value of pi with formatted string literals**

```python
from math import pi
print('This is the value of pi', pi)
```

**Example 5. Representing pi with a precision of 2 decimal values**

```python
print(f"The value of pi is approximately {pi:.4f}")
```

```python
number = 18.91018190
print(f'{number: .2f}')
```

> Note: that the first `f` is related to the formatted string. The `:.2f` means that the value of pi is formatted as a decimal with exactly two digits.

## 2. String `format()` Method

* The `str.format()` method of strings requires more manual effort.
* We will still use `{` and `}` to mark where a variable will be substituted and can provide detailed formatting directives, but we also need to provide the information to be formatted.

**Example 1. Student taking a course without any formatting**

```python
name = "Bob"
subject = "Algorithms"
year = "2024"
print("The student", name, "is taking the course", subject, "in", year)
```

**Example 2. Printing with a formatted string literal**

```python
print(f"The student {name} is taking the course {subject} in {year}")
```

**Example 3. Printing using string fomat**

```python
print("The student {} is taking the course {} in {}".format(name, subject, year))
```

**Example 4. Printing fruits in order**

```python
print("{0}, {1} and {2}".format('apple', 'banana', 'watermelon'))
```

**Example 5. Printing fruits in different order**

```python
print("{2}, and {0}".format('apple', 'banana', 'watermelon'))
```

**Example 6. Aligning to the left, right, and center**

```python
# < ===> alinear a la izquierda
# > ===> alinear a la derecha
# ^ ===> alinear al centro
print("{:<100}".format("Valentina"))
print("{:>100}".format("Chelsea"))
print("{:^100}".format("Juan Carlos"))
```

> Note that the number 10 specifies the total width that will be used and then it aligns to the left, right, or center respectively, but using 10 characters width.

**Example 7. Counting total of votes**

```python
votes = 91471947190
null_votes = 71481478
percentage_of_votes = null_votes/votes
print('From a total of {:,} votes, the {:.2%} of the votes were null votes'.format(votes, percentage_of_votes))
```

```python
area = 14141973.9208802178716
print('El area es: {:,.3f}'.format(area))
```

## 3. Manual string formatting

We can use some methods to manually format strings:
* `str.rjust()`
* `str.ljust()`
* `str.center()`

The `str.rjust()` method right-justifies a string in a field of a given width by padding it with spaces on the left. The methods `str.ljust()` and `str.center()` are similar.

**Example 1. Right-Justyfing different names**

```python
name1 = "Ben"
name2 = "Oliver"
name3 = "Maximiliano"
```

```python
print(name1.rjust(15))
print(name2.rjust(15))
print(name3.rjust(15))
```

**Example 2. Left-Justyfing different countries**

```python
country1 = "Switzerland"
country2 = "Mexico"
country3 = "United States of America"
```

```python
print(country1.ljust(15))
print(country2.ljust(15))
print(country3.ljust(15))
```

**Example 3. Center-Justyfing different fruits**

```python
fruit1 = "apple"
fruit2 = "strawberry"
fruit3 = "watermelon"
```

```python
print(fruit1.center(30))
print(fruit2.center(30))
print(fruit3.center(30))
```

* When reading from a file, we can use the reserved keyword `with` to open the file. This is followed by the function `open()`, which requires two arguments: 1) the name of the file, and 2) the mode in which we would like to open the file (e.g., 'r' for reading).
* Using `with` ensures that the file is properly closed after its suite finishes, even if an exception is raised.

* There are three methods that can be used to read a file:
1. `read()`: Reads the entire contents of the file.
2. `readline()`: Reads one line at a time from the file.
3. `readlines()`: Reads all the lines of the file and returns them as a list.

**Example 1. Using `read()`**

```python
print("10\n8\n7")
```

```python
with open('files/scores.txt', 'r') as f:
  file = f.read()
  print(file)
```

```python
with open ('files/scores.txt', 'r') as f:
    line = f.readline().strip()
    while line:
        print(line)
        line = f.readline().strip()
```

**Example 3. Using `readlines()`**

```python
with open('files/scores.txt', 'r') as f:
  scores = f.readlines()
  print(scores)
```

## 5. Writing to file

* To write to a file, we can use the same reserved keyword `with`, followed by the function `open()`. Inside the parentheses, we specify the name of the file and the mode in which we want to open the file. 

* There are two primary modes for writing to a file:
1. 'w' mode: Opens the file for writing. If the file already exists, it will be overwritten. If the file does not exist, it will be created.
2. 'a' mode: Opens the file for appending. New data will be added at the end of the file without deleting the existing content. If the file does not exist, it will be created.

**Example 1. Writing into a file named finalSales**

```python
with open('finalSales.txt', 'w') as f:
    f.write('1100\n')
    f.write('20\n')
    f.write('7999\n')
```

**Example 2. Appending into a file named finalSales**

```python
with open('finalSales.txt', 'a') as f:
    f.write('23\n')
    f.write('77\n')
    f.write('83\n')
```

**4. Write a program that reads student names and their grades from a file, calculates the average grade, and displays the result with only two decimal places in a new file.**

```python
# names_and_grades = ['Juan Carlos, 10.0', 'Marcela, 8.4', 'Luisa, 9.1', 'Fernando, 7.8', 'Mateo, 8.8']
# 10.0, 8.4, 9.1, 7.8, 8.8
with open('Files/students_and_grades.txt', 'r') as file:
  document = file.read() 
  print(document)
  '''
  document = 'Juan Carlos, 10.0\nMarcela, 8.4\nLuisa, 9.1\nFernando, 7.8\nMateo, 8.8'
  '''
  names_and_grades = document.split('\n')
  # names_and_grades = ['Juan Carlos, 10.0', 'Marcela, 8.4', 'Luisa, 9.1', 'Fernando, 7.8', 'Mateo, 8.8']
  suma = 0
  for element in names_and_grades:
    # name_and_grade = ['Juan Carlos', '10.0']
    name_and_grade = element.split(',')
    print(name_and_grade)
    grade = float(name_and_grade[1])
    # grade = 10.0
    suma = suma + grade
  average = suma / len(names_and_grades)
  #print(f"The average of the grades is: {average:.2f}")
  print("The average of the grades is: {:.2f}".format(average))

with open('Files/result.txt', 'w') as write_file:
  result = "The average of the grades is: {:.2f}".format(average)
  print(result)
  write_file.write(result)
  
  
  
```
