```python
'''
1. Contar del 1 al 10

Descripción: Crear un diagrama de flujo que muestre los números del 1 al 10 utilizando un ciclo.
Ejemplo: Mostrar "1, 2, 3, ..., 10".
'''

for numero in range(1, 11, 1):
  print(numero)
```

```python
numero = 1
while numero <= 10:
  print(numero)
  numero += 1
```

```python
'''2. Factorial de un número

Descripción: Crear un diagrama de flujo que lea un número y calcule su factorial utilizando un ciclo.
Ejemplo: Si el número es 4, mostrar "El factorial de 4 es 24".

'''

numero = int(input("Dame un numero"))
factorial = 1
if numero == 0:
  print("El factorial de cero es 1")
elif numero < 0:
  print("No tiene factorial")
else:
  i = 1
  while i <= numero:
    factorial = factorial * i
    i += 1
  print(numero, "tiene como factorial", factorial)
```

```python
'''
3. Tabla de multiplicar

Descripción: Crear un diagrama de flujo que lea un número y muestre su tabla de multiplicar hasta 10.
Ejemplo: Si el número es 3, mostrar "3x1=3, 3x2=6, ..., 3x10=30".
'''

numero = int(input("Dame un numero"))
print("tabla del", numero)
for i in range(1, 11, 1):
  resultado = numero * i
  #print(numero, "x", i, "=", resultado)
  print(f'{numero} x {i} = {resultado}')

```
